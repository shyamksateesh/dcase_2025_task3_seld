## Doing Distance Normalization

### First, we grab all the distance values

```python
import glob
import os
import pandas as pd
from rich.progress import Progress
label_files = glob.glob(os.path.join("DCASE2025_SELD_dataset", 'metadata_dev', 'dev-*', '*.csv'))

all_distances = []

with Progress() as progress:
    task = progress.add_task("[cyan]Processing label files...", total=len(label_files))
    for file in label_files:
        df = pd.read_csv(file)
        raw_distance = (df['distance'] / 100.0).tolist() 
        all_distances.extend(raw_distance)
        progress.update(task, advance=1)
```

### Plot the distribution of distances

```python
import matplotlib.pyplot as plt
import numpy as np 


# Plotting the histogram of distances.
plt.figure(figsize=(8, 6))
plt.hist(all_distances, bins=30, edgecolor='black', alpha=0.7)
plt.xlabel('Distance (meters)')
plt.ylabel('Frequency')
plt.title('Distribution of Distances')
plt.grid(True)
plt.show()
```

### Now, let's extract dmin, dstd, and dmax that are used for the `parameters.py` file

```python
# get d_mean and d_std
mean_dist = np.mean(all_distances)
std_dist = np.std(all_distances)

d_stand = (all_distances - mean_dist)/std_dist

# get d_max
d_max = max(d_stand)
d_norm = d_stand/max(d_stand)

# These are the values that we will use for the parameters.py file
print(f"Mean: {mean_dist}, Std: {std_dist}, Max: {d_max}")

# Plotting the normalized distances histogram.
plt.figure(figsize=(8, 6))
plt.hist(d_norm, bins=80, edgecolor='black', alpha=0.7)
plt.xlabel('Normalized Distance [-1, 1]')
plt.ylabel('Frequency')
plt.title('Distribution of Normalized Distances')
plt.grid(True)
plt.show()
```
